```python
import os
import requests
import json
import sys
from openai import OpenAI

# Get openai key
# Use current working directory and go one level up
parent_dir = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.append(parent_dir)

# Now you can import your config
from config import api_key
```

```python
# Set the prompt
prompt = "To succeed in life "
```

```python
completion_params = {
    "model": "davinci-002",
    "prompt": prompt,   # prompt is just a string
    "max_tokens": 50
}

# Initialize the OpenAI API client
client = OpenAI(api_key=api_key)

# Call the OpenAI API to generate a completion
response = client.completions.create(**completion_params)

# Extract the generated completion from the API response
completion_text = response.choices[0].text.strip()

print("Generated completion:", prompt + completion_text)
```

```python
chat_params = {
    "model": "gpt-3.5-turbo",  # or gpt-4o etc.
    "messages": [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ],
    "max_tokens": 50
}

# Initialize the OpenAI API client
client = OpenAI(api_key=api_key)

response = client.chat.completions.create(**chat_params)
completion_text = response.choices[0].message.content.strip()
print("Generated completion:", completion_text)
```

```python
print(response.model_dump())
```
